```python
import pandas as pd

df = pd.read_csv('../data/netflix_titles.csv')
print("Dataset loaded successfully!")
```

```python
#standardizing column names, by lowercasing them and removing space

df.columns = df.columns.str.strip().str.lower().str.replace(' ', '_')
```

```python
print("Missing values before cleaning...")
print(df.isnull().sum().to_string())
print(f"\nInitial row count: {len(df)}")
```

```python
duplicate_count = df.duplicated().sum()
if duplicate_count > 0:
    df = df.drop_duplicates() 
    print(f" Removed {duplicate_count} duplicate rows.")
else:
    print(" No duplicate rows detected.")
```

```python
text_columns_to_fill = ['director', 'cast', 'country']
df[text_columns_to_fill] = df[text_columns_to_fill].fillna('Unknown')

df['rating'] = df['rating'].fillna('Unknown')

if 'date_added' in df.columns:
    df['date_added'] = df['date_added'].str.strip()

df = df.dropna(subset=['date_added', 'duration'])
print("Missing values handled.")
```

```python
df['date_added'] = pd.to_datetime(df['date_added'], format='mixed')
df['date_added'] = df['date_added'].dt.strftime('%d-%m-%Y') 
print(" Date formats standardized to dd-mm-yyyy.")
print("\nRandom sample of formatted dates:")
print(df['date_added'].sample(10).to_string())
```

```python
print(df.isnull().sum().to_string())
print(f"\nFinal cleaned row count: {len(df)}")
```

```python
df.to_csv('../data/cleaned_dataset.csv', index=False)
print(" Success! The clean, structured dataset has been exported.")
```
